```python
# 3.Write a program that catches IndexError when accessing a list element.
def access_element(lst, index):
    try:
        print(f"Element at index {index}: {lst[index]}")
    except IndexError:
        print("Error: Index out of range.")

# Example usage
numbers = [10, 20, 30, 40, 50]

print("List:", numbers)
idx = int(input("Enter the index to access: "))

access_element(numbers, idx)



```

```python
# 4.Write a program to open a file and handle FileNotFoundError.
def read_file(filename):
    try:
        with open(filename, "r") as file:
            content = file.read()
            print("File content:\n", content)
    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found.")

# Example usage
filename = input("Enter filename to open: ")
read_file(filename)
```

```python
# 10.Write a program that raises and handles a TypeError.
def add_numbers(a, b):
    try:
        if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
            # Raising TypeError manually
            raise TypeError("Both inputs must be numbers.")
        result = a + b
    except TypeError as e:
        print("TypeError Caught:", e)
    else:
        print("Result:", result)

# Example usage
add_numbers(10, 5)      # ✅ Works fine
add_numbers(10, "abc")  # ❌ Raises and handles TypeError
```

```python
# 11.Write a function to withdraw money and raise an exception if balance < withdrawal.
class InsufficientBalanceError(Exception):
    """Custom exception for insufficient balance."""
    pass

def withdraw(balance, amount):
    try:
        if amount > balance:
            raise InsufficientBalanceError("Withdrawal amount exceeds current balance.")
        balance -= amount
    except InsufficientBalanceError as e:
        print("Transaction Failed:", e)
    else:
        print(f"Withdrawal successful! Remaining balance: {balance}")
    return balance

# Example usage
balance = 5000
balance = withdraw(balance, 2000)  # ✅ Works
balance = withdraw(balance, 4000)  # ❌ Raises custom exception
```

```python
# 12.Write a function that validates age and raises an exception if age < 18.
class UnderAgeError(Exception):
    """Custom exception for underage users."""
    pass

def validate_age(age):
    try:
        if age < 18:
            raise UnderAgeError("Age must be at least 18.")
        print("Age is valid ✅ You are eligible.")
    except UnderAgeError as e:
        print("Validation Failed:", e)

# Example usage
validate_age(20)
validate_age(15)


```

```python
# 14.Write a custom exception NegativeNumberError and use it in a square root function.
import math

class NegativeNumberError(Exception):
    """Custom exception for negative numbers in square root."""
    pass

def safe_sqrt(num):
    try:
        if num < 0:
            raise NegativeNumberError("Square root of negative number is not allowed.")
        result = math.sqrt(num)
    except NegativeNumberError as e:
        print("Error:", e)
    else:
        print(f"Square root of {num} = {result}")

# Example usage
safe_sqrt(25)
safe_sqrt(-9)

```

```python
# 15.Write a custom exception for InsufficientFundsError in a banking system.
class InsufficientFundsError(Exception):
    """Custom exception for insufficient funds in banking transactions."""
    pass

class BankAccount:
    def __init__(self, balance=0):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited: {amount}  | New Balance: {self.balance}")

    def withdraw(self, amount):
        try:
            if amount > self.balance:
                raise InsufficientFundsError("Insufficient funds for this withdrawal.")
            self.balance -= amount
            print(f"Withdrew: {amount} | Remaining Balance: {self.balance}")
        except InsufficientFundsError as e:
            print("Transaction Failed:", e)

# Example usage
account = BankAccount(5000)
account.deposit(2000)     #  Balance = 7000
account.withdraw(3000)    #  Balance = 4000
account.withdraw(6000)    #  Raises and handles InsufficientFundsError
```

```python
# 16.Write a program that handles exceptions while parsing JSON.
import json

def parse_json(data):
    try:
        result = json.loads(data)
    except json.JSONDecodeError as e:
        print("Error: Invalid JSON format →", e)
    else:
        print("Valid JSON  Parsed data:", result)

# Example usage
valid_json = '{"name": "Alice", "age": 25}'
invalid_json = '{"name": "Alice", "age": 25'   # Missing closing brace

parse_json(valid_json)    #  Works fine
parse_json(invalid_json)  #  Caught JSONDecodeError
```

```python
# 17.Write a program to retry dividing two numbers 3 times if an error occurs.
def safe_division():
    attempts = 3
    for i in range(attempts):
        try:
            num1 = int(input("Enter numerator: "))
            num2 = int(input("Enter denominator: "))
            result = num1 / num2
        except ValueError:
            print("Error: Please enter valid integers.")
        except ZeroDivisionError:
            print("Error: Division by zero is not allowed.")
        else:
            print(f"Result: {result}")
            break
    else:
        print("❌ Maximum attempts reached. Try again later.")

# Example usage
safe_division()
```

```python
# 18.Write a program that uses assert to validate positive numbers and handle AssertionError.
def validate_positive(num):
    try:
        assert num > 0, "Number must be positive."
    except AssertionError as e:
        print("AssertionError Caught:", e)
    else:
        print(f"Valid number ✅: {num}")

# Example usage
validate_positive(10)   # Valid
validate_positive(-5)   # AssertionError
```

```python
# 20.Write a program that raises multiple custom exceptions (InvalidPasswordError, InvalidUsernameError).
class InvalidUsernameError(Exception):
    """Custom exception for invalid username."""
    pass

class InvalidPasswordError(Exception):
    """Custom exception for invalid password."""
    pass

def login(username, password):
    try:
        if username != "admin":
            raise InvalidUsernameError("Invalid username. Access denied.")
        if password != "12345":
            raise InvalidPasswordError("Invalid password. Try again.")
        print("✅ Login successful!")
    except InvalidUsernameError as e:
        print("Login Failed:", e)
    except InvalidPasswordError as e:
        print("Login Failed:", e)

# Example usage
login("admin", "12345")     #  Success
login("user", "12345")      #  InvalidUsernameError
login("admin", "wrong")     #  InvalidPasswordError
```
